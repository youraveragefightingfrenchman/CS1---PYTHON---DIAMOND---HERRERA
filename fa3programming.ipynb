```python
name = input("Enter student's name: ")
score = float(input("Enter student's score: "))

if 90 <= score <= 100:
  grade = "A"
elif 80 <= score <= 89:
  grade = "B"
elif 70 <= score <= 79:
  grade = "C"
elif score <70:
  grade = "F"

print(f"\nStudent Name: {name}")
print(f"Score: {score}")
print(f"Letter Grade: {grade}")

```
